## Q6. Median horsepower, before and after fillna

1. Median of `horsepower`.
2. Most frequent value of `horsepower`.
3. `fillna` the missing values with that most frequent value.
4. Median again — has it changed? *(increased / decreased / no)*

> Work on a copy, not on `df` itself, so the earlier answers stay reproducible.
> The real lesson here: imputing with the mode drags the distribution toward the mode. Ask yourself which direction it *should* move before you compute it.


```python
median_before = df['horsepower'].median()
mode = df['horsepower'].mode()[0]
hp_filled = df['horsepower'].fillna(mode)
median_after = hp_filled.median()
print(median_before, mode, median_after)
```

## Q7. Sum of weights

1. Cars from Asia only
2. Columns `vehicle_weight` and `model_year`
3. First 7 rows
4. `X = ` the underlying NumPy array
5. `XTX = X.T @ X`
6. Invert `XTX`
7. `y = [1100, 1300, 800, 900, 1000, 1100, 1200]`
8. `w = inv(XTX) @ X.T @ y`
9. Sum of the elements of `w`? *(0.0369 / 0.369 / 3.69 / 36.9)*

> This is the normal equation: `w = (XᵀX)⁻¹ Xᵀ y`. You are fitting a least-squares linear regression by hand, two lessons before they name it.
> Same move as solving a small linear system in a simulation calibration — the inverse exists only because the 7×2 slice gives you a well-conditioned 2×2.


```python
carsFromAsia = df[df['origin'] == 'Asia']
weightAndYear = carsFromAsia[['vehicle_weight', 'model_year']]
X = weightAndYear.head(7).values
XTX = X.T @ X
Xinv = np.linalg.inv(XTX)
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
w = Xinv @ X.T @ y
wsum = w.sum()
wsum
```
